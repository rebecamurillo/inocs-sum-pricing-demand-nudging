```python
import json, pandas as pd, re

L_cap = 10  # cap you want to check against

def load_df(solution_path):
    with open(solution_path, "r", encoding="utf-8") as f:
        sol = json.load(f)
    df = pd.DataFrame(sol.get("instances", []))

    # ---- OD parsing (keep your robust fallbacks) ----
    if "OD" in df.columns and df["OD"].notna().any():
        def parse_OD(s):
            s = str(s)
            m = re.split(r"\s*[-–—>→]\s*", s)
            if len(m) >= 2 and m[0] != "" and m[1] != "":
                return int(float(m[0])), int(float(m[1]))
            return s, None
        df[["o","d"]] = df["OD"].apply(parse_OD).apply(pd.Series)
    elif "od" in df.columns:
        def parse_od(s):
            try:
                o, d = str(s).split("_", 1)
                return int(float(o)), int(float(d))
            except Exception:
                return s, None
        df[["o","d"]] = df["od"].apply(parse_od).apply(pd.Series)
    elif all(c in df.columns for c in ["o","d"]):
        pass
    else:
        raise ValueError("Could not infer OD columns from the JSON schema.")

    # Normalize type (optional)
    if "Type" in df.columns:
        df["Type"] = df["Type"].astype(str).str.upper()

    # Sanity: ensure int dtype when possible
    for c in ["o","d"]:
        if pd.api.types.is_numeric_dtype(df[c]):
            df[c] = df[c].astype("Int64")

    return df

def analyze(solution_path):
    df = load_df(solution_path)

    # ---- OD universe from the data (solution-independent) ----
    od_universe = df[["o","d"]].dropna().drop_duplicates()
    od_universe["__key__"] = 1  # helper for merges/joins if needed

    # If nu column exists, compute *offered/selected* counts; else fill zeros
    if "nu" in df.columns:
        offered = df[df["nu"] == 1].copy()
        offered_counts = (
            offered.groupby(["o","d"]).size()
            .rename("n_total")
            .reset_index()
        )
    else:
        offered_counts = pd.DataFrame(columns=["o","d","n_total"])

    # Merge counts onto the fixed OD universe; fill missing with 0
    res = (
        od_universe.merge(offered_counts, on=["o","d"], how="left")
        .fillna({"n_total": 0})
        .astype({"n_total": int})
    )

    # Optional split by Type on the offered subset (kept separate, does not affect OD count)
    if "Type" in df.columns and "nu" in df.columns:
        split = (
            df[df["nu"] == 1]
            .pivot_table(index=["o","d"], columns="Type", aggfunc="size", fill_value=0)
            .reset_index()
        )
        res = res.merge(split, on=["o","d"], how="left").fillna(0).rename_axis(None, axis=1)

    # Metrics that are stable in OD count, variable in offered arcs
    avg_arcs_per_od = res["n_total"].mean()
    n_exceed_L = (res["n_total"] > L_cap).sum()
    n_exceed_18 = (res["n_total"] > 18).sum()

    summary = {
        "File": solution_path,
        "Total ODs (fixed by data)": int(len(od_universe)),
        "ODs seen with nu==1": int((res["n_total"] > 0).sum()),
        "Average arcs per OD (nu==1)": round(float(avg_arcs_per_od), 3),
        f"ODs with total > {L_cap}": int(n_exceed_L),
        "ODs with total > 18": int(n_exceed_18),
    }
    return summary, res

# Example usage across files
for p in [
    r"C:\Users\mkoseogl\Desktop\Geneva Data\relatota\instos_and_sols\instanto\sols_int_slack\strongduos\margarita\durge_pup\Instance_1669_fix_euclid_0.5_None_invDefault_solution.json",
    r"C:\Users\mkoseogl\Desktop\Geneva Data\relatota\instos_and_sols\instanto\sols_int_slack\strongduos\margarita\durge_pup\Instance_5870_fix_euclid_0.5_None_invDefault_solution.json",
]:
    summary, _ = analyze(p)
    display(summary)
```
